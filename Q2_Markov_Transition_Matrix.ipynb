Taking input from the user

```python
input_seq = input("Enter the input sequence for calculation of matrix : ")
print("Entered sequence : "+input_seq)

for i in range(len(input_seq)):
    if input_seq[i] not in ['A','C','G','T']:
        print("You entered invalid sequence")
        break

```

Calculating frequencies of all dinucleotides in the input sequence

```python
dinucleotide_count={}

for i in range(len(input_seq)-1):
    first_char=input_seq[i]
    second_char=input_seq[i+1]
    dinucleotide=first_char+second_char
    if dinucleotide in dinucleotide_count:
        dinucleotide_count[dinucleotide]+=1
    else:
        dinucleotide_count[dinucleotide]=1
```

Calculating probabilities of each dinucleotide seen in the input sequence

```python
for key,value in dinucleotide_count.items():
    dinucleotide_count[key]=value/(len(input_seq)-1)
```

Using conditional probabilty to find markov matrix

```python
bases=['A','C','G','T']

markov_matrix=[]

for first_base in bases:
    probabilities=[]
    sum=0
    for second_base in bases:
        dinucleotide=first_base+second_base
        if dinucleotide in dinucleotide_count:
            sum+=dinucleotide_count[dinucleotide]
    
    for second_base in bases:
        dinucleotide=first_base+second_base
        if dinucleotide in dinucleotide_count:
            probabilities.append(dinucleotide_count[dinucleotide]/sum)
        else:
            probabilities.append(0)
    
    markov_matrix.append(probabilities)
```

Printing Final Transition Matrix

```python
print("Transition matrix : ")
print("","A","C","G","T",sep="\t")

for i in range(4):
    print(bases[i],end="\t")
    for j in range(4):
        print(f"{markov_matrix[i][j]:.3f}",end="\t")
    print()

    
```

Reading the file and concatenating the related lines under one section

```python
def convert_fasta_to_single_line(input_file, output_file):
    with open(input_file, 'r') as infile, open(output_file, 'w') as outfile:
        sequence = ''
        for line in infile:
            line = line.strip()
            if line.startswith('>'):
                if sequence:
                    outfile.write(sequence + '\n')
                    sequence = ''
                outfile.write(line + '\n')
            else:
                sequence += line
        if sequence:
            outfile.write(sequence + '\n')
```

```python
convert_fasta_to_single_line("multiline_input.fasta","output_file.txt")
```
